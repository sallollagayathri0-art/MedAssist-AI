```python
def get_risk_level(score):
    if score <= 4:
        return "Low"
    elif score <= 7:
        return "Medium"
    else:
        return "High"


print("Risk level function created successfully!")
```

```python
def calculate_risk_score(age, blood_pressure, cholesterol, severity):
    score = 0

    # Age risk
    if age >= 60:
        score += 3
    elif age >= 40:
        score += 2
    elif age >= 30:
        score += 1

    # Blood pressure risk
    if blood_pressure == "High":
        score += 3
    elif blood_pressure == "Normal":
        score += 1

    # Cholesterol risk
    if cholesterol == "High":
        score += 3
    elif cholesterol == "Normal":
        score += 1

    # Symptom severity
    if severity == "High":
        score += 3
    elif severity == "Medium":
        score += 2
    else:
        score += 1

    return score


print("Risk scoring function created successfully!")
```

```python
def calculate_severity(fever, cough, fatigue, difficulty_breathing):
    score = fever + cough + fatigue + difficulty_breathing

    if score <= 1:
        return "Low"
    elif score <= 2:
        return "Medium"
    else:
        return "High"


print("Severity function created successfully!")
```
